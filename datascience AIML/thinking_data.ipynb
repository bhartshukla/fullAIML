## caleaning and structuring our data


```python
# clean and structure our data
# data is incorrect missing data , mixed data  
# duplicate data 
def cleane_data(data):
    text_to_number = {"one":1, "two":2, "three":3, "four":4, "five":5}
    
    cleanddata = []
    unique_user = set()
    
    
    for user in data:
        # print(user)
      
        # clean data
        raw_rating = str(user["rating"]).strip().lower()
        # print(raw_rating)   # ✅
        if(raw_rating in text_to_number):
            raw_rating = text_to_number[raw_rating]

        user["rating"] = raw_rating
        
# print(data)      

        # handle missing value |
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None


        # Deduplication
        if(user["name"].strip() in unique_user):
           continue

        unique_user.add(user["name"])
        cleanddata.append(user)


    return cleanddata   


            
# print (data)

 ## getting meaningful insights form data 
   ## whats avg rating 
    ## whats % of usres gave poor rating (<3)
    
        

```

```python
data = cleane_data(data)
print(data)
```

## getting meaningful insights from data 
## whats avg rating
## what% of user gave poor rating

```python
def get_insights(data):

    
    tot_rating  =  0
    for user in data:
        tot_rating += float(user["rating"])
        
        
    print(f"avg_rating = {tot_rating/len(data)}")

    # poor rating 

    poor_rating = 0
    for user in data:
        if(float(user["rating"])<5):
            poor_rating += 1

    print(f"poor rating % = {poor_rating/len(data)*100}")
 



    
    
```

```python
get_insights(data)
```

## Build a product recommendation feature

```python
# if user reating >=4 , recommend same barnd products
# if user reating < t , recommend diff barand products 

def recomdation(data):

    brand = {
        "rating_more_then_4":"apple",
        "rating_less_then_4" :"lava"
    }

    for user in data:
        if(float(user["rating"]) >= 4):
            print(f"{user["name"]} you can chose {brand["rating_more_then_4"]} product")

        if(float(user["rating"]) < 4):
            print(f"{user["name"]} you can chose  {brand["rating_less_then_4"]} product")
            
        
            
```
